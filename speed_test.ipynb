```python
# python zip
l1 = [3,4,5]
l2 = [6,7,8]
l3 = [8,9,1]
list(zip(l1,l2,l3))
```

```python
# using lists
import numpy as np
import time

size = 1_000_000
l1 = list(range(size))
l2 = list(range(size))

start = time.time()
add = [x+y for x,y in zip(l1,l2)]
end = time.time()
print(add[0:10])
print(end - start)
```

```python
# using numpy arrays
import numpy as np
import time

size = 1_000_000
l1 = np.array(list(range(size)))
l2 = np.array(list(range(size)))

start = time.time()
add = l1 + l2
end = time.time()
print(add[0:10])
print(end - start)
```
